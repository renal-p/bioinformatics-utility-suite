```python
import math
import itertools
```

```python
# simple perm calculator

def solve_perm(n):
    # 1. Calculate total number of permutations (n!)
    total = math.factorial(n)
    print(total)
    
    # 2. Generate the base sequence: [1, 2, ..., n]
    elements = range(1, n + 1)
    
    # 3. Stream and print the permutations directly
    # Using a generator avoids loading massive lists into memory
    for p in itertools.permutations(elements):
        print(*p)

# Example usage for Rosalind input (e.g., n = 3)
solve_perm(3)
```

```python
# perm calculator but with directions

def oriented_permutations(n):
    # 1. Create the pool of all possible directed elements: [-2, -1, 1, 2]
    pool = list(range(-n, 0)) + list(range(1, n + 1))

    # 2. Get all length-n combinations from this pool
    all_combinations = itertools.permutations(pool, n)

    # 3. Filter the set: Only keep combinations where all absolute values are unique
    # In symbolic logic: ∀ x, y in permutation (x ≠ y ∧ |x| ≠ |y|)
    oriented_permutations = [
        p for p in all_combinations 
        if len(set(abs(x) for x in p)) == n
    ]
    return oriented_permutations
# Print the results
results = oriented_permutations(2)
print(len(results))
for value in results:
    print(*value)
```
