```python
import pandas as pd

# Convert your list of aligned DNA sequences directly into a tabular grid
sequence_list = ["ATCCGCTTAG", "TCCGCTTAG", "ATCCGATT_G", 'TGCTATGAT'] # your raw input strings

# Finds the length of the smallest DNA sequence
min_length = min(len(seq) for seq in sequence_list)

# Ensures all sequences have the same length as the shortest sequence, by truncating at the same nucleotide number
truncated_sequences = [seq[:min_length] for seq in sequence_list]

# Each row is a sequence, each column is a nucleotide position (0, 1, 2...)
df_sequences = pd.DataFrame([list(seq) for seq in truncated_sequences])

# Labels the columns starting with 1
df_sequences.columns = range(1, len(df_sequences.columns) + 1)

# Compute positional counts across columns instantly using .apply() and value_counts()
# Fill missing rows (like a missing 'G' at a position) with 0, and ensure data reads as integers
profile_df = df_sequences.apply(lambda col: col.value_counts()).fillna(0).astype(int)

# Restructure index to guarantee standard ACGT row ordering
profile_df = profile_df.reindex(['A', 'C', 'G', 'T'], fill_value=0)

# Extract the single consensus string using idxmax() along the vertical axis (axis=0)
# This automatically grabs the character ('A','C','G','T') with the highest frequency per column
consensus_string = "".join(profile_df.idxmax(axis=0))

print(consensus_string)
print(profile_df)
```
