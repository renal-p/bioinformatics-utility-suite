```python
def rabbit_population(initial_conditions, n, k):
    # initial conditions i.e. [F(n), F(n-1)]
    state_vector = list(initial_conditions)
    for month in range(1, n):
        #State vector loaded into memory: index 0 = adults, index 1 = offspring
        current_adults = state_vector[0]
        current_offspring = state_vector[1]
       
        #Row 1 calculation: [1,k] dot [current_adults, current_offspring] = F(n+1)
        next_adults = (1 * current_adults) + (k * current_offspring)
        
        # Row 2 calculation: [1,0] dot [current_adults, current_offspring] = F(n), becomes new baseline
        next_offspring = (1 * current_adults) + (0 * current_offspring)

        #Packages F(n+1) and F(n) into new [F(n), F(n-1)]
        state_vector = [next_adults, next_offspring]
    return state_vector[0]

final_pairs = rabbit_population([1,0], 5, 3)
print('Final Rabbit Pairs:', final_pairs)
```
